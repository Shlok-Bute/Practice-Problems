```python
dict_ = {0:90 , 1:80 , 2:70 , 3:40 , 4:50 , 5:20 , 6:30}
sort = []
key = []

for i in dict_.values() :
    sort.append(i)

for i in dict_.keys() :
    key.append(i)   
    
temp = 0
temp2 = 0

for i in range(len(key)) :
    for k in range (i , len(key)) :
        if(sort[i] > sort[k]):
            temp = sort[i]
            sort[i] = sort[k]
            sort[k] = temp
            temp2 = key[i]
            key[i] = key[k]
            key[k] = temp2
            

sort_dict = {}

for i in range (len(key)):
    sort_dict[key[i]] = sort[i]

print(sort_dict)    
```
