```python
numbers = (10,20,30,40,50,60,70,80,90)

print("4th element from beginning " , numbers[3])
print("4th element from end " , numbers[-4])

num = int(input("Enter number to search"))

if num in numbers :
    print("Element present")
else :
    print("Element not present")

```

```python
lst = [1,2,3,4,5,6]
tupl = tuple(lst) 
print(tupl)

num = int(input("Enter number"))

if num in tupl :
    print("Index of element " , tupl.index(num))
else :
    print("Element not found")
    
```

```python
lst = [(10,20,30) , (40,50,60) , (70,80,90)]
new_lst = []

for tup in lst :
    new_tup = tup[:-1] + (100,)
    new_lst.append(new_tup)

print(new_lst)    
```

```python
tup = (1,10,6,3,7,10,-1)
minm = tup[0]
maxm = tup[0]
for i in tup :
    if(minm > i):
        minm = i
    if(maxm < i):
        maxm = i
        
print("max " , maxm)
print("min " , minm)
```
