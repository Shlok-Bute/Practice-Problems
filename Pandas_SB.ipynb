```python
import pandas as pd

a = [1,2,3,4]
dic = {"100" : 1 , "101" : 2 , "102" : 3 , "103" : 4}

tabl = pd.Series(a , index = ['a' , 'b' , 'c' , 'd'] , name ="Example Series")
tabl2 = pd.Series(dic , name = "Dictionary table")

print(tabl)
print(end = "\n")
print(tabl2)
```

```python
import pandas as pd

data = {"cal" : [123,145,175] , "dur" : [30,24,64]}

table = pd.DataFrame(data ,  index = ["day1","day2","day3"])

print(table.loc[["day1","day2"]])
print(end ='\n')
print(table)
```

```python
import pandas as pd

data = {
    'A' : [1,2,3,None,5],
    'B' : [None,2,3,4,5],
    'C' : [1,2,3,None,5],
    'D' : [1,2,3,4,None],
}

df = pd.DataFrame(data)
df2 = pd.DataFrame(data)
df3 = pd.DataFrame(data)

print(df)
print(end = "\n")
df.fillna(0 , inplace=True)
print(df)
print(end = "\n")
print(df2.dropna())
print(end = "\n")
df3.fillna(df3.mean() , inplace=True)
print(df3)
```
