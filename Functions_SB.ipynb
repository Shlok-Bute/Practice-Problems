```python
def fun(counter,cnt,counter2,case) :
    i = 0
    j = 0
    k = 0
    
    while i<cnt :
        while j<cnt :
            if(j>=counter) :
               print("*" , end='')
            else :
               print(" " , end='')
            j+=1
        while k<counter2 :
            print("*" , end = '')
            k+=1
            
        j = 0
        k = 0
        i+=1
        counter-=case
        counter2+=case
        print()

fun(3,4,0,1)
fun(1,3,3,-1)
```

```python
def triangle (a,b,c) :
    
    if a**2 == b**2 + c**2 :
        print("Right angled triangle")
    else :
        print("Not a right angled triangle")

d  = int(input("enter  number"))
e  = int(input("enter  number"))  
f  = int(input("enter  number")) 


if d > e and d > f:
    triangle(d,e,f)
elif e > d and e > f:
    triangle(e,d,f)
else:
    triangle(f,d,e)
```

```python
def sum_product(a,b,c):
    sum2 = a+b+c
    prod2 = a*b*c
    return sum2 , prod2

d  = int(input("enter  number"))
e  = int(input("enter  number"))  
f  = int(input("enter  number")) 

sum1 , prod1 = sum_product(d,e,f)
print(sum1 ," ", prod1)
```
