```python
lst = [10,20,30,40,50]

num = 0
i = len(lst)-1
j = 0 
while i!=j:
    num = lst[j]
    lst[j] = lst[i]
    lst[i] = num
    j+=1
    i-=1

print("reversed list " , lst)
```

```python
lst = [10,20,30,20,50,10,30,10]
num = int(input("Enter number to search"))
occur = 0
for i in range(len(lst)):
    if(lst[i] == num):
        occur+=1

print(num , "occurs" , occur , "times in the list")
        
```

```python
lst = [10,20,30,40,50]
choice = int(input("press 1 to remove by element number or 2 to remove by index"))

if choice == 1 :
    num = int(input("enter number to remove in list"))
    lst.remove(num) #re,oving by element
elif choice == 2 :    
    pos = int(input("enter at which position to remove : list length = 5"))
    lst.pop(pos-1) #removing by index
else:
    print("invalid choice")

print(lst)
```
