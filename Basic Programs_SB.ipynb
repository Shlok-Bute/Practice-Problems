```python
marks = float(input("enter marks"))

if marks >= 90.0 :
    print("Grade --> 'O'")
elif marks>=80.0 : 
    print("Grade --> 'A+'")
elif marks>=70.0 :
    print("Grade --> 'A'")
elif marks>=60.0 :
    print("Grade --> 'B'")    
elif marks>=50.0 :
    print("Grade --> 'C'")
elif marks>=40.0 :
    print("Grade --> 'P'")
else :
    print("Grade --> 'F'")
```

```python
num = int(input("Enter number"))

sum = 0

while num>0 :
    sum+=num%10
    num = num//10

print("sum of digits is " , sum)
    
```

```python
num = int(input("enter a number"))
fact = 1

for i in range(1,num+1):
    fact*=i

print("factorial = " ,fact)    
```

```python
num = int(input("enter a number"))
factors = 0

for i in range (2,int(num**0.5)+1):
    if num%i == 0 :
        factors+=1

if(factors==0):
    print("prime")
else:
    print("not prime")
```

```python
year = int(input("enter year"))

if year%400==0:
    print("leap year")
elif year%4==0 and year%100!=0:
    print("leap year")
else:
    print("not leap")
```
