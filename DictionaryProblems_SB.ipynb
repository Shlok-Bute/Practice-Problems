```python
dict = {0:10 , 1:20 , 2:30 , 3:40 , 4:50 , 5:60 , 6:70}

for keys in dict :
    print(keys)

print("---------")

for val in dict.values() :
    print(val)

print("---------")
for k,v in dict.items() :
    print(k , ":" , v)
```

```python
dict_ = {}

for i in range(1 , 16) :
    dict_[i] = i**2

print(dict_)

sum = 0

for i in dict_.values() :
    sum+=i

print(sum)
```
